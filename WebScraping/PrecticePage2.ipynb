```python
items=[]
for article in articles:
    title=article.find('h3').find('a')['title']
    price=article.select_one("p.price_color").text.split('£')[1]
    rating_element=article.select_one("p.star-rating")
    ratting=rating_element["class"][1]
    availability = soup.find_all("p", class_="instock availability")
    for item in availability:
        avail=(item.get_text(strip=True))
    items.append([title,price,ratting,avail])
    
```

```python
import pandas as pd 
```

```python
df=pd.DataFrame(items,columns=['Book Title','Book Price','Book Rating',"Stcok"])
df
```
