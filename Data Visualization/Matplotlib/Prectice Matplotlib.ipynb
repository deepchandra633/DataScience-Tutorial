```python
import matplotlib.pyplot as plt
plt.style.use('classic')

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.plot(years, runs,)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
import numpy as np
x=np.arange(len(years))
import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))  # index positions
width = 0.25

plt.bar(x-width,sachin,width=width,label="Sachin",color='red',ls='--')
plt.bar(x,sehwag,width=width,label="Sehwag",color='green',ls=':')
plt.bar(x+width,kohli,width=width,label="Kohli",color='blue',ls='-.')
plt.xlabel("Years")
plt.ylabel("Runs Scored")
plt.title("Run Comparasion")
plt.xticks(x,years)
plt.legend(loc="upper left")
plt.show()
```

```python
players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.barh(players, runs, color="green")

# Add labels at the end of each bar
for i in range(len(players)):
    plt.text(runs[i] + 10, i, str(runs[i]), va='center')

plt.xlabel("Runs")
plt.ylabel("Players")
plt.title("Runs Scored by Players")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]  # Days of the week
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5]

labels = ['Studying', 'Playing', 'Watching TV', 'Sleeping']
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']

plt.figure(figsize=(10,6))
plt.stackplot(days, studying, playing, watching_tv, sleeping,
              labels=labels,colors=colors, alpha=0.8,edgecolor='black')

plt.legend(loc='upper right')  # Location of the legend
plt.title('Weekly Activity Tracker')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.grid(True,alpha=1)
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]

sales = [120,180,160,210,250]

plt.figure(
    figsize=(9,5),
    dpi=150,
    facecolor="lightyellow"
)

plt.plot(
    months,
    sales,
    marker="o",
    color="blue",
    markerfacecolor="red",
    markeredgecolor="black",
    linewidth=3,
    label="Sales"
)

plt.title(
    "Monthly Sales Report",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Months")

plt.ylabel("Sales")

plt.grid(
    linestyle="--",
    alpha=0.5
)

plt.legend()

plt.savefig("sales_report.png", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(
    x,
    y,
    marker="o",
    color="blue",             # Line color
    markerfacecolor="red",    # Marker ke andar ka color
    markeredgecolor="black",  # Marker ki boundary ka color
    markeredgewidth=2,         # Boundary ki thickness
    markersize=10             # Marker ka size
)

plt.text(3, 30, "Middle Point")

plt.show()
```
