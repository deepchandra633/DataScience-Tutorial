```python
import matplotlib.pyplot as plt

# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 95, 50, 55, 20, 65, 15, 85, 90]

plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
color=['red' if score<50 else 'green' for score in exam_scores]
size=[score*2 for score in exam_scores]
plt.scatter(study_hours, exam_scores,c=color,s=size)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python

plt.scatter(study_hours, exam_scores, c=exam_scores, cmap='viridis')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
plt.scatter(study_hours, exam_scores, c=exam_scores, cmap='viridis')
plt.colorbar(label='Score')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
score=[100,200,300,400,500,600,700,800,900]
plt.scatter(study_hours, exam_scores, c=score, cmap='cool')
plt.colorbar(label='Score')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
plt.scatter(study_hours, exam_scores, c=score, cmap='cool')
for i in range(len(study_hours)):
    plt.annotate(f'S {i+1}', (study_hours[i]+0.1, exam_scores[i]))
plt.colorbar(label='Score')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
# Assume two groups: Class A and Class B
class_a_hours = [2, 4, 6, 8]
class_a_scores = [45, 55, 65, 85]

class_b_hours = [1, 3, 5, 7, 9]
class_b_scores = [40, 50, 60, 70, 90]

plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue')
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange')

plt.title('Scatter Plot: Class A vs Class B')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

class_a_hours = [2, 4, 6, 8]
class_a_scores = [45, 55, 65, 85]
class_b_hours = [1, 3, 5, 7, 9]
class_b_scores = [40, 50, 60, 70, 90]


scatter_a = plt.scatter(class_a_hours, class_a_scores, label='Class A', c=class_a_scores, cmap='cool', marker='X', s=100, edgecolor='black')
scatter_b = plt.scatter(class_b_hours, class_b_scores, label='Class B', c=class_b_scores, cmap='cool', marker='o', s=100, edgecolor='black')


plt.colorbar(scatter_a, label="Score Intensity_A")
plt.colorbar(scatter_b, label="Score Intensity_B")

plt.title('Scatter Plot: Class A & Class B Analysis')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5,6,7]
marks = [30,40,45,60,65,75,90]

plt.figure(figsize=(8,5))

plt.scatter(
    hours,
    marks,
    color="blue",
    s=120,
    marker="o",
    edgecolors="black",
    alpha=0.8
)

plt.title("Study Hours vs Marks")

plt.xlabel("Study Hours")

plt.ylabel("Marks")

plt.grid(True)

plt.show()
```
