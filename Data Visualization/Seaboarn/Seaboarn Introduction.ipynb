```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="ticks") 
```

```python
x = np.array([0, 2, 3, 4, 5, 6, 7, 8, 9 ,10, 60])
y = np.sin(x)

sns.lineplot(x=x, y=y)
plt.title('Beautiful Line Plot')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
plt.figure(figsize=(10,5))

df = pd.DataFrame({
    "Student":["A","B","C","D","E"],
    "Marks":[70,80,65,90,85]
})
plt.subplot(1,2,1)
sns.barplot(
    data=df,
    x="Student",
    y="Marks",
    palette="viridis",
    hue="Marks"
)
plt.subplot(1,2,2)
sns.lineplot(
    data=df,
    x="Student",
    y="Marks"
)
plt.tight_layout()


plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Student":["A","B","C","D","E","F","G","H"],
    "Gender":["Male","Female","Male","Female","Male","Male","Female","Male"]
})


sns.countplot(
    data=df,
    x="Gender",
    hue="Gender",
    palette="Set2",
    legend=False,
    # order=["Female","Male"]
)
plt.show()
```
