```python
import matplotlib.pyplot as plt

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
```

```python
plt.plot(years, runs,)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
plt.bar(years, runs,)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))  # index positions
width = 0.25

plt.bar(x-width,sachin,width=width,label="Sachin")
plt.bar(x,sehwag,width=width,label="Sehwag")
plt.bar(x+width,kohli,width=width,label="Kohli")
plt.xlabel("Years")
plt.ylabel("Runs Scored")
plt.title("Run Comparasion")
plt.xticks(x,years)
plt.legend()
```

```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color="skyblue")

# Add labels on top of bars
for i in range(len(players)):
    plt.text(i, runs[i] + 30, str(runs[i]), ha='center')

plt.ylabel("Runs")
plt.xlabel("Players")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```

```python
# create a side by side bar chart comparing runs of rohit virat rahul over 5 selected years
import numpy as np
import matplotlib.pyplot as plt

years = ["2018", "2019", "2020", "2021", "2022"]

rohit = [1030, 1490, 842, 906, 995]
virat = [2735, 2455, 842, 964, 1348]
rahul = [1251, 1105, 401, 952, 821]

x = np.arange(len(years))
width = 0.2

plt.figure(figsize=(10, 6))

# Bars
plt.bar(x - width, rohit, width, label="Rohit Sharma", color="skyblue")
plt.bar(x, virat, width, label="Virat Kohli", color="orange")
plt.bar(x + width, rahul, width, label="KL Rahul", color="green")

plt.title("Runs Comparison of Rohit, Virat and KL Rahul")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.xticks(x, years)

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend()
# plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

players = ["Rohit", "Virat", "Rahul"]

total_runs = [18000, 28000, 12000]

plt.figure(figsize=(15,5))

plt.barh(players, total_runs,
         color=["skyblue","orange","green"])

# Add labels
for i in range(len(players)):
    plt.text(total_runs[i]+50,
             i,
             total_runs[i],
             va="center")

plt.xlabel("Runs")
plt.ylabel("Players")
plt.title("Total Runs in Debut 5 Years")

plt.grid(axis="x",alpha=0.5)

plt.show()
```
