```python
import matplotlib.pyplot as plt

ages = [
    34, 28, 36, 45, 27, 27, 45, 37, 25, 35,
    25, 25, 32, 10, 12, 24, 19, 33, 20, 15,
    44, 27, 30, 15, 24, 31, 18, 33, 23, 27,
    23, 48, 29, 19, 38, 17, 32, 10, 16, 31,
    37, 31, 28, 26, 15, 22, 25, 40, 33, 12,
    33, 26, 23, 36, 40, 39, 21, 26, 33, 39,
    25, 28, 18, 18, 38, 43, 29, 40, 33, 23,
    33, 45, 29, 45, 3, 38, 30, 27, 30, 10,
    27, 33, 44, 24, 21, 24, 39, 33, 24, 35,
    30, 39, 22, 26, 26, 15, 32, 32, 30, 27
]

bins = [10, 20, 30, 40, 50, 60, 70]

plt.hist(ages, bins=bins, edgecolor='black')
plt.title('Age Distribution of YouTube Viewers')
plt.xlabel('Age Group')
plt.ylabel('Number of Viewers')
plt.grid(True, linestyle='--', alpha=0.5)
# plt.show()
```

```python
plt.hist(ages, bins=bins, edgecolor='black')
plt.axvline(22,color='red',linestyle='--',lw=3,label="average")
plt.title('Age Distribution of YouTube Viewers')
plt.xlabel('Age Group')
plt.ylabel('Number of Viewers')
plt.grid(True, linestyle='--', alpha=0.5)
# plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45,48,50,52,55,56,60,62,65,68,
         70,72,74,75,78,80,82,85,88,90]

plt.hist(marks,bins=5,edgecolor="red",linestyle='--',color="yellow")
plt.grid(True)
plt.show()
```

```python
plt.hist(marks,bins=5,edgecolor="red",linestyle='--',color="yellow",orientation="horizontal")
plt.grid(True)
plt.show()
```

```python

plt.hist(marks,bins=5,edgecolor="red",linestyle='--',color="yellow",density=True)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [
10,48,50,52,15,56,
60,22,65,68,
40,32,74,45,
78,80,52,85,
28,90
]
bins=[10,30,50,70,90]
plt.figure(figsize=(8,5))

plt.hist(
    marks,
    bins=bins,
    color="skyblue",
    edgecolor="black",
    alpha=0.8,
    rwidth=0.9
)

plt.title("Distribution of Marks")

plt.xlabel("Marks")

plt.ylabel("Number of Students")

plt.grid(axis="y")

plt.show()
```
