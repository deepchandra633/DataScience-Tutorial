```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")  # Choose any style you like
```

```python
# Data
Players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=Players)
plt.show()
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=Players, colors=colors)
plt.show()
```

```python
plt.pie(
    runs,
    labels=Players,
    colors=colors,
    wedgeprops={'edgecolor': 'green', 'linewidth': 2, 'linestyle': '-.'}
)
plt.show()
```

```python
# fill
plt.pie(
    runs,
    labels=Players,
    colors=colors,
    wedgeprops={'edgecolor': 'green', 'linewidth': 2, 'linestyle': '-.','fill':False}
)
plt.show()
```

```python
explode = [0.2, 0, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs, labels=Players, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
explode = [0.2, 0, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs, labels=Players, explode=explode, colors=colors,shadow=True)
plt.title("Exploded Pie Example")
plt.show()
```

```python
explode = [0.2, 0, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs,
        labels=Players, 
        explode=explode, 
        colors=colors,shadow=True,
        startangle=100
       )
plt.title("Exploded Pie Example")
plt.show()
```

```python
explode = [0.2, 0, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs,
        labels=Players, 
        explode=explode, 
        colors=colors,shadow=True,
        startangle=100,
        autopct='%1.1f%%'
       )
plt.title("Exploded Pie Example")
plt.show()
```

```python
# create a pie chart showing the market share of mobile os(Andriod,ios,others).then re create the same using
# a horizontal bar chart and observe which is easy to  understand

import matplotlib.pyplot as plt

mobile_os = ["Android", "iOS", "Others"]
market_share = [72, 27, 1]

plt.figure(figsize=(6,6))

plt.pie(
    market_share,
    labels=mobile_os,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.2,0,0],
    colors=["green", "silver", "orange"],
    shadow=True
)

plt.title("Mobile OS Market Share")

plt.show()
```

```python
import matplotlib.pyplot as plt

mobile_os = ["Android", "iOS", "Others"]
market_share = [72, 27, 1]

plt.figure(figsize=(10,5))

plt.barh(
    mobile_os,
    market_share,
    color=["green", "blue", "orange"]
)

# Add labels
for i in range(len(mobile_os)):
    plt.text(
        market_share[i] + 1,
        i,
        f"{market_share[i]}%",
        va="center"
    )

plt.title("Mobile OS Market Share")

plt.xlabel("Market Share (%)")

plt.ylabel("Operating System")

plt.grid(axis="x")

plt.show()
```
